### --- Hands-on Exercise 2: Education Analytics - Student Performance Analysis


```python
import numpy as np

# Create the scores array with the provided dataset
scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91])

print("Student Scores:", scores)
print("Data Type of Array:", scores.dtype)
print("Shape of Array:", scores.shape)
```

```python
# Calculate the average score
mean_score = np.mean(scores)

print(f"Group Mean Score: {mean_score}")
```

```python
# Calculate individual score deviations from the mean
deviations = scores - mean_score

print("Individual Deviations from Mean:")
print(deviations)
```

### Interpretation of Deviations
* **Positive Deviation:** Indicates that the student scored *above* the classroom average. They are performing better than the group baseline.
* **Negative Deviation:** Indicates that the student scored *below* the classroom average. These students may need additional targeted learning interventions.
* **Value Close to Zero:** Indicates that the student's performance perfectly aligns with the classroom baseline average.


```python
# Square the deviations
squared_deviations = deviations ** 2

print("Squared Deviations:")
print(squared_deviations)
```

### Why Squaring Changes Negative Values to Positive
Squaring a number involves multiplying it by itself. According to basic algebraic rules, multiplying two negative numbers yields a positive product (for example: $-5 \times -5 = 25$). Mathematically, this eliminates negative signs so that negative performance gaps do not cancel out positive performance gaps when aggregated together.


```python
# Calculate the total of the squared deviations
sum_squared_deviations = np.sum(squared_deviations)

print(f"Sum of Squared Deviations: {sum_squared_deviations:.2f}")
```

### The Standard Deviation Conceptual Process
1. **Calculate the mean:** Find the central benchmark of the scores.
2. **Calculate deviations:** Determine how far each individual score sits from that benchmark.
3. **Square the deviations:** Eliminate negative signs so differences do not sum to zero.
4. **Sum the squared deviations:** Total up the overall variations.
5. **Continue with the appropriate division:** Divide by the number of samples ($N$) to find the variance (the average squared distance).
6. **Take the square root:** Pull the units back to the original scale (test scores instead of squared test scores).

### Why Standard Deviation is Useful in Education Analytics
Standard deviation tells us the *consistency* of student learning. A low standard deviation means most students score close to the average, proving the material is universally absorbed. A high standard deviation warns us that the class is highly polarized-meaning some students excel while others are heavily struggling—which prompts immediate instructional adjustments.
