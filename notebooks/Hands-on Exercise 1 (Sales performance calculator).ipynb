# Hands-On Exercise 1: Business Analytics — Sales Performance Calculator

##  Problem Scenario
For this project I am supporting the analytics team of a growing retail business. The management team wants to analyze a week's worth of daily sales data to understand overall performance, calculate baseline metrics, and simulate the financial impact of a hypothetical 10% sales increase.

##  Objectives
In this notebook, I will use NumPy to:
1. Represent raw numerical sales data as a 1D array and inspect its structure.
2. Compute descriptive metrics like total and average weekly revenue.
3. Perform element-wise scalar multiplication and vector subtraction to simulate revenue growth.
4. Translate data outcomes into a business-focused executive summary.

```python
### --- Task 1: Creating dataset
import numpy as np
daily_sales = np.array([125000,150000,175000,140000,190000,210000,160000])

print("Daily Sales:", daily_sales)
print("Data Type of Array:", daily_sales.dtype)
print("Shape of the Array:", daily_sales.shape)
```

```python
### --- Task 2: Calculate total sales
total_sales = np.sum([daily_sales])
print(f"Total Sales:", total_sales)
```

```python
### --- Task 3: Calculate Average Daily Sales
average_sales = np.average([daily_sales])
print(f"Average Sales:", average_sales)
```

```python
### --- Task 4: Apply a 10% increase using element-wise scalar multiplication
adjusted_sales = daily_sales * 1.10

print("Adjusted Sales (with 10% increase):")
print(adjusted_sales)
```

```python
### --- Task 5: Calculate the daily revenue difference
sales_difference = adjusted_sales - daily_sales

print("Additional Revenue Earned Per Day:")
print(sales_difference)
```

#### --- The total sales figure tells the business the absolute volume of revenue generated over the entire week, which is crucial for overall financial tracking and matching high-level targets.In contrast, the average daily sales figure provides a baseline measure of typical daily performance. It irons out day-to-day fluctuations, giving management a reliable benchmark to forecast future daily cash flow, evaluate staffing needs, and quickly identify which specific days underperformed or overperformed against the norm.

#### --- Executive Briefing: 
Weekly Sales Performance:
This business generated a strong total revenue of $1,151,000 over the past week, maintaining a healthy daily average of $164,428.57.
