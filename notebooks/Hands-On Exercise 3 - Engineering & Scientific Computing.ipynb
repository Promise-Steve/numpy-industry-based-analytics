# Hands-On Exercise 3: Engineering & Scientific Computing - Trigonometric Series and Convergence

####  Problem Scenario
I am supporting an engineering/scientific computing team working with a mathematical series. The team wants to understand how a series approximation behaves as more terms are included in a calculation. This notebook uses NumPy to generate numerical sequences, evaluate trigonometric fractions, and observe the concept of convergence.

####  Objectives
In this notebook, I will use NumPy to:
1. Convert angles from degrees to radians using `np.radians()`.
2. Generate continuous discrete tracking sequences using `np.arange()`.
3. Leverage broadcasting to perform element-wise array division (\(\frac{\sin(\theta)}{k}\)).
4. Calculate and compare progressive partial sums to observe numerical convergence.

```python
import numpy as np

# Define angle in degrees
theta = 30

# Convert angle to radians
theta_radians = np.radians(theta)

print(f"Angle in Degrees: {theta}")
print(f"Angle in Radians: {theta_radians:.6f}")
```

```python
# Create sequence for k from 1 up to (but excluding) 100
k = np.arange(1, 100, 1)

print("Generated k array:\n", k)
print("\n--- Array Inspection ---")
print("First value:", k[0])
print("Last value:", k[-1])
print("Number of values generated:", len(k))
```

```python
# Calculate sine of the radian angle
sin_component = np.sin(theta_radians)

print(f"sin({theta}°) component value: {sin_component}")
```

```python
# Create individual terms using NumPy broadcasting (scalar / vector)
series_terms = sin_component / k

# Calculate the sum of the generated terms
total_sum = np.sum(series_terms)

print("First 5 terms of the series:", series_terms[:5])
print(f"Sum of terms (for k=1 to 99): {total_sum:.6f}")
```

```python
# Evaluate the series across different range thresholds
sum_99 = np.sum(sin_component / np.arange(1, 100, 1))
sum_999 = np.sum(sin_component / np.arange(1, 1000, 1))
sum_9999 = np.sum(sin_component / np.arange(1, 10000, 1))

print("Progressive range evaluations completed successfully.")
```

### Convergence Investigation Table

| Number of Terms | Series Result |
| :--- | :--- |
| **99** | 2.588689 |
| **999** | 3.742235 |
| **9,999** | 4.893753 |

#### Observation:
As the number of terms increases, the result continues to grow, but the *rate* at which it grows slows down dramatically. Each subsequent order of magnitude adds a smaller fractional contribution to the total running sum.


### What Convergence Means in This Experiment

* **Fewer Terms:** When using only 99 terms, we get a rough, early approximation. The baseline calculation leaves out thousands of tiny subsequent fractions.
* **Increased Terms:** Moving to 9,999 terms includes values where $k$ is very large. This makes individual elements ($\frac{\sin(\theta)}{k}$) incredibly close to zero.
* **Why the Approximation Changes:** Each added term expands our precision. Because the denominator ($k$) grows continually while the numerator ($\sin(30^\circ) = 0.5$) stays completely constant, the terms being added shrink closer and closer to zero.
* **Limiting Value Meaning:** In scientific computing, approaching a limiting value means the algorithm is stabilizing. Even though we are adding thousands of new inputs, they become too small to significantly shift our primary calculations. This allows engineers to safely stop calculating once a desired level of numerical precision is reached.
